```python


def keep_number_unit(text):
    if text is None:
        return None
    
    match = re.search(r'[\d.]+\s*[kL]', text)
    return match.group().replace(" ", "") if match else None
```

```python
name=[]
rating=[]
review=[]
salaries=[]
interviews=[]
jobs=[]
benefits=[]
highly_rated_for=[]
critically_rated_for=[]

industry=[]
headquarter=[]
other_locations_count=[]
```

```python
import re

for i in company:

    # --- NAME ---
    h2 = i.find('h2')
    name.append(h2.text.strip() if h2 else None)

    # --- RATING ---
    r = i.find('div', class_='rating_text')
    rating.append(r.text.strip() if r else None)

    # --- INDUSTRY | HEADQUARTER | OTHER LOCATIONS ---
    span = i.find('span', class_='companyCardWrapper__interLinking')
    
    if span:
        text = span.text.strip()

        parts = text.split('|')
        industry.append(parts[0].strip())

        location_part = parts[1].strip() if len(parts) > 1 else None

        if location_part:
            # HQ
            hq = re.search(r'^[A-Za-z /]+', location_part)
            headquarter.append(hq.group().strip() if hq else None)

            # Other locations count
            count = re.search(r'\+(\d+)', location_part)
            other_locations_count.append(int(count.group(1)) if count else 0)
        else:
            headquarter.append(None)
            other_locations_count.append(0)
    else:
        industry.append(None)
        headquarter.append(None)
        other_locations_count.append(0)

    # --- HIGHLY & CRITICALLY RATED FOR ---
    rating_values_element = i.find_all('span', class_='companyCardWrapper__ratingValues')

    highly_rated_for.append(
        rating_values_element[0].text.strip() if len(rating_values_element) > 0 else None
    )

    critically_rated_for.append(
        rating_values_element[1].text.strip() if len(rating_values_element) > 1 else None
    )

    # --- ACTION BUTTONS ---
    action = i.find_all('a', class_='companyCardWrapper__ActionWrapper')

    review.append(action[0].text.strip() if len(action) > 0 else None)
    salaries.append(action[1].text.strip() if len(action) > 1 else None)
    interviews.append(action[2].text.strip() if len(action) > 2 else None)
    jobs.append(action[3].text.strip() if len(action) > 3 else None)
    benefits.append(action[4].text.strip() if len(action) > 4 else None)
```

```python
d = {
    'name': name,
    'rating': rating,
    'industry': industry,
    'headquarter': headquarter,
    'other_locations': other_locations_count,
    'highly_rated_for': highly_rated_for,
    'critically_rated_for': critically_rated_for,
    'review': review,
    'salaries': salaries,
    'interviews': interviews,
    'jobs': jobs,
    'benefits': benefits
}

df = pd.DataFrame(d)
df
```

```python
import re
import time
import requests
from bs4 import BeautifulSoup

name=[]
rating=[]
reviews=[]
salaries=[]
interviews=[]
jobs=[]
benefits=[]
highly_rated_for=[]
critically_rated_for=[]
industry=[]
headquarter=[]
other_locations_count=[]

headers = {
    'User-Agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0 Safari/537.36'
}

for j in range(1, 501):

    print(f"Scraping page {j}...")

    url = f'https://www.ambitionbox.com/list-of-companies?page={j}'
    response = requests.get(url, headers=headers, timeout=10)

    if response.status_code != 200:
        print(f"⚠️ Page {j} failed")
        continue

    soup = BeautifulSoup(response.text, 'html.parser')
    companies = soup.find_all('div', class_='companyCardWrapper')

    for i in companies:

        # ---------- NAME ----------
        h2 = i.find('h2')
        name.append(h2.text.strip() if h2 else None)

        # ---------- RATING ----------
        r = i.find('div', class_='rating_text')
        rating.append(r.text.strip() if r else None)

        # ---------- INDUSTRY | HQ | LOCATIONS ----------
        span = i.find('span', class_='companyCardWrapper__interLinking')

        if span:
            text = span.text.strip()
            parts = text.split('|')

            industry.append(parts[0].strip())

            location_part = parts[1].strip() if len(parts) > 1 else None

            if location_part:
                hq = re.search(r'^[A-Za-z /]+', location_part)
                headquarter.append(hq.group().strip() if hq else None)

                count = re.search(r'\+(\d+)', location_part)
                other_locations_count.append(int(count.group(1)) if count else 0)
            else:
                headquarter.append(None)
                other_locations_count.append(0)
        else:
            industry.append(None)
            headquarter.append(None)
            other_locations_count.append(0)

        # ---------- HIGHLY / CRITICALLY RATED ----------
        rating_values = i.find_all('span', class_='companyCardWrapper__ratingValues')

        highly_rated_for.append(
            rating_values[0].text.strip() if len(rating_values) > 0 else None
        )

        critically_rated_for.append(
            rating_values[1].text.strip() if len(rating_values) > 1 else None
        )

        # ---------- ACTION BUTTONS (FIXED) ----------
        action_dict = {
            'Reviews': None,
            'Salaries': None,
            'Interviews': None,
            'Jobs': None,
            'Benefits': None
        }

        actions = i.find_all('a', class_='companyCardWrapper__ActionWrapper')

        for a in actions:
            title = a.find('span', class_='companyCardWrapper__ActionTitle')
            count = a.find('span', class_='companyCardWrapper__ActionCount')

            if title and count:
                key = title.text.strip()
                if key in action_dict:
                    action_dict[key] = count.text.strip()

        reviews.append(action_dict['Reviews'])
        salaries.append(action_dict['Salaries'])
        interviews.append(action_dict['Interviews'])
        jobs.append(action_dict['Jobs'])
        benefits.append(action_dict['Benefits'])

    # 🔒 Delay to avoid blocking
    time.sleep(1)
```

```python
import pandas as pd

df = pd.DataFrame({
    'name': name,
    'rating': rating,
    'industry': industry,
    'headquarter': headquarter,
    'other_locations': other_locations_count,
    'highly_rated_for': highly_rated_for,
    'critically_rated_for': critically_rated_for,
    'reviews': reviews,
    'salaries': salaries,
    'interviews': interviews,
    'jobs': jobs,
    'benefits': benefits
})

df.isna().sum()
```

```python
df['reviews'] = df['review'].apply(keep_number_unit)
df['salaries'] = df['salaries'].apply(keep_number_unit)
df['interviews'] = df['interviews'].apply(keep_number_unit)
df['jobs'] = df['jobs'].apply(keep_number_unit)
df['benefits'] = df['benefits'].apply(keep_number_unit)
```
